### Ejercicio 7 · Scope

Sin ejecutar la celda, ¿qué imprimirá? Asigna tu predicción a `prediccion` ("10", "20" o "error") y luego ejecuta para comprobar.

```python
# === EJERCICIO 7 ===
prediccion = None  # "10", "20" o "error"

x = 10
def cambiar():
    x = 20   # crea una variable LOCAL nueva, no toca la de fuera
cambiar()
print(x)

assert prediccion == "10", "La x de dentro es local: la de fuera sigue valiendo 10"
print("✅ ¡Correcto!")
```
