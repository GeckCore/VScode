### Ejercicio 7 · Dataclass

Convierte esta clase en una `@dataclass` y observa que todo sigue funcionando con mucho menos código.

```python
# === EJERCICIO 7 ===
from dataclasses import dataclass

# convierte esto a dataclass:
class Estudiante:
    def __init__(self, nombre, expediente, nota_media):
        self.nombre = nombre
        self.expediente = expediente
        self.nota_media = nota_media

e1 = Estudiante("Ana", "A1", 8.5)
e2 = Estudiante("Ana", "A1", 8.5)
assert e1.nombre == "Ana"
# si tu dataclass está bien hecha, esto será True (comparación automática):
assert e1 == e2
print(e1)
print("✅ ¡Correcto!")
```
