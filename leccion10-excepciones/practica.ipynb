### Ejercicio 3 · else y finally

```python
# === EJERCICIO 3 ===
def division_segura(a, b):
    registro = "inicio"
    try:
        r = a / b
    except ZeroDivisionError:
        mensaje = "falló"
    # ← completa con else y finally
    return mensaje, registro

assert division_segura(10, 2) == ("ok: 5.0", "fin")
assert division_segura(1, 0) == ("falló", "fin")
print("✅ ¡Correcto!")
```
