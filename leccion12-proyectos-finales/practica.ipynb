## 🥇 Proyecto C: Inventario RPG (POO)

Completa las clases. Los tests validan la lógica; luego hay una demo jugable.

```python
from dataclasses import dataclass

@dataclass
class Objeto:
    nombre: str
    peso: float
    valor: int

class Inventario:
    def __init__(self, peso_maximo):
        self.peso_maximo = peso_maximo
        self.objetos = []

    def peso_actual(self):
        """Suma de pesos de los objetos."""
        pass

    def añadir(self, objeto):
        """Añade si cabe por peso. Devuelve True si se añadió, False si no cabía."""
        pass

    def objeto_mas_valioso(self):
        """Devuelve el objeto con más valor, o None si no hay."""
        pass

    def __str__(self):
        lineas = [f"- {o.nombre} ({o.peso} kg, {o.valor} oro)" for o in self.objetos]
        return "Inventario:\n" + "\n".join(lineas)
```

```python
# TESTS del Proyecto C (no toques esta celda)
inv = Inventario(peso_maximo=10)
assert inv.añadir(Objeto("Espada", 3, 100)) is True
assert inv.añadir(Objeto("Escudo", 4, 80)) is True
assert inv.añadir(Objeto("Armadura", 5, 300)) is False  # no cabe: 3+4+5 > 10
assert inv.peso_actual() == 7
assert inv.objeto_mas_valioso().nombre == "Espada"
assert Inventario(5).objeto_mas_valioso() is None
print(inv)
print("\n✅ Proyecto C superado. Bonus: añade guardado/cargado JSON (dicts con dataclasses.asdict).")
```
