### Ejercicio 8 · lru_cache

Compara la velocidad de Fibonacci recursivo con y sin `@lru_cache`. Ejecuta y observa los tiempos.

```python
from functools import lru_cache
import time

def fib_lento(n):
    if n < 2:
        return n
    return fib_lento(n - 1) + fib_lento(n - 2)

@lru_cache(maxsize=None)
def fib_rapido(n):
    if n < 2:
        return n
    return fib_rapido(n - 1) + fib_rapido(n - 2)

t = time.time(); fib_lento(30); print(f"Sin caché: {time.time()-t:.3f}s")
t = time.time(); fib_rapido(30); print(f"Con caché: {time.time()-t:.6f}s")
```

### Ejercicio 9 · Reto: decorador contador

Escribe el decorador `contar_llamadas` que añada a la función un atributo `.llamadas` con cuántas veces se ha ejecutado.

```python
# === EJERCICIO 9 ===


@contar_llamadas
def hola():
    return "hola"

hola(); hola(); hola()
assert hola.llamadas == 3
print("✅ ¡Correcto!")
```
